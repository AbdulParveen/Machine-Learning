```python
#Total Budget Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Total_Budget_Billions_USD"], bins=30)
plt.title("Distribution of Total Budget")
plt.show()
```

```python
#Budget Trend by Year
plt.figure(figsize=(12,5))
sns.lineplot(x="Year", y="Total_Budget_Billions_USD", data=df)
plt.title("Budget Trend Over Years")
plt.show()
```

```python
#Top 10 Countries by Average Budget
top = df.groupby("Country")["Total_Budget_Billions_USD"].mean().sort_values(ascending=False).head(10)

top.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Countries by Average Budget")
plt.ylabel("Budget (Billions USD)")
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(14,10))

sns.heatmap(df.select_dtypes(include=np.number).corr(),
            annot=False,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
